1.Identify  top spending customers

```python
top_customers = pd.read_sql("""
SELECT Customer_ID,
       SUM(Final_Amount) AS total_spent,
       COUNT(Order_ID) AS total_orders
FROM online_food_orders
GROUP BY Customer_ID
ORDER BY total_spent DESC
LIMIT 10
""", engine)

top_customers
```

```python


plt.figure(figsize=(10,5))

plt.bar(
    top_customers['Customer_ID'],
    top_customers['total_spent']
)

plt.title("Top 10 Spending Customers")
plt.xlabel("Customer ID")
plt.ylabel("Total Amount Spent")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Revenue & Profit Analysis


4.Monthly revenue trends


```python
monthly_revenue_df = pd.read_sql("""
SELECT 
    DATE_FORMAT(Order_Date, '%%Y-%%m') AS order_month,
    SUM(Final_Amount) AS total_revenue
FROM online_food_orders
GROUP BY order_month
ORDER BY order_month
""", engine)

monthly_revenue_df
```

```python
monthly_revenue_df['order_month'] = pd.to_datetime(monthly_revenue_df['order_month'])


plt.figure(figsize=(8,4))
plt.plot(monthly_revenue_df['order_month'],
         monthly_revenue_df['total_revenue'],
         marker='o')

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

5.Impact of discounts on profit

```python
discount_profit_df = pd.read_sql("""
SELECT
    CASE
        WHEN Discount_Applied = 0 THEN 'No Discount'
        WHEN Discount_Applied <= 50 THEN 'Low Discount'
        WHEN Discount_Applied <= 150 THEN 'Medium Discount'
        ELSE 'High Discount'
    END AS discount_category,
    AVG(Profit_Margin_Percent) AS avg_profit_margin,
    COUNT(Order_ID) AS total_orders
FROM online_food_orders
GROUP BY discount_category
ORDER BY avg_profit_margin DESC
""", engine)

discount_profit_df
```

```python

plt.figure(figsize=(7,4))

plt.bar(
    discount_profit_df['discount_category'],
    discount_profit_df['avg_profit_margin']
)

plt.title("Impact of Discounts on Profit Margin")
plt.xlabel("Discount Category")
plt.ylabel("Average Profit Margin (%)")

plt.tight_layout()
plt.show()
```

Delivery Performance

7.Average delivery time by city


```python
avg_delivery_city_df = pd.read_sql("""
SELECT
    City,
    AVG(Delivery_Time_Min) AS avg_delivery_time,
    COUNT(Order_ID) AS total_deliveries
FROM online_food_orders
WHERE Delivery_Time_Min IS NOT NULL
  AND Order_Status = 'delivered'
GROUP BY City
ORDER BY avg_delivery_time DESC
""", engine)

avg_delivery_city_df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9,4))

plt.bar(
    avg_delivery_city_df['City'],
    avg_delivery_city_df['avg_delivery_time']
)

plt.xticks(rotation=45)
plt.title("Average Delivery Time by City")
plt.xlabel("City")
plt.ylabel("Average Delivery Time (Minutes)")

plt.tight_layout()
plt.show()
```

8.Distance vs delivery delay analysis

```python
distance_delay_df = pd.read_sql("""
SELECT
    CASE
        WHEN Distance_km <= 2 THEN '0-2 km'
        WHEN Distance_km <= 5 THEN '2-5 km'
        WHEN Distance_km <= 10 THEN '5-10 km'
        ELSE '10+ km'
    END AS distance_bucket,
    AVG(Delivery_Time_Min) AS avg_delivery_time,
    COUNT(Order_ID) AS total_orders
FROM online_food_orders
WHERE Delivery_Time_Min IS NOT NULL
  AND Order_Status = 'delivered'
GROUP BY distance_bucket
ORDER BY distance_bucket;
""", engine)

distance_delay_df
```

```python


plt.figure(figsize=(7,4))

plt.bar(
    distance_delay_df['distance_bucket'],
    distance_delay_df['avg_delivery_time'],
    color='skyblue'
)

plt.title("Average Delivery Time by Distance")
plt.xlabel("Distance Bucket (km)")
plt.ylabel("Average Delivery Time (Minutes)")

plt.tight_layout()
plt.show()
```

9.Delivery rating vs delivery time

```python
rating_delay_df = pd.read_sql("""
SELECT
    Delivery_Rating,
    AVG(Delivery_Time_Min) AS avg_delivery_time,
    COUNT(Order_ID) AS total_orders
FROM online_food_orders
WHERE Delivery_Rating IS NOT NULL
  AND Delivery_Time_Min IS NOT NULL
  AND Order_Status = 'delivered'
GROUP BY Delivery_Rating
ORDER BY Delivery_Rating;
""", engine)

rating_delay_df
```

Restaurant Performance


10.Top-rated restaurants


```python
top_restaurants_df = pd.read_sql("""
SELECT
    Restaurant_Name,
    AVG(Restaurant_Rating) AS avg_rating,
    COUNT(Order_ID) AS total_orders
FROM online_food_orders
WHERE Restaurant_Rating IS NOT NULL
GROUP BY Restaurant_Name
HAVING total_orders >= 50
ORDER BY avg_rating DESC
LIMIT 10;
""", engine)

top_restaurants_df
```

```python


plt.figure(figsize=(10,5))

plt.bar(
    top_restaurants_df['Restaurant_Name'],
    top_restaurants_df['avg_rating'],
    color='orange'
)

plt.title("Top-Rated Restaurants")
plt.xlabel("Restaurant Name")
plt.ylabel("Average Rating")
plt.xticks(rotation=45, ha='right')
plt.ylim(0, 5)  # Ratings max is 5
plt.tight_layout()
plt.show()
```

11.Cancellation rate by restaurant

```python
cancellation_df = pd.read_sql("""
SELECT
    Restaurant_Name,
    COUNT(CASE WHEN Order_Status = 'cancelled' THEN 1 END) AS cancelled_orders,
    COUNT(Order_ID) AS total_orders,
    (COUNT(CASE WHEN Order_Status = 'cancelled' THEN 1 END) / COUNT(Order_ID)) * 100 AS cancellation_rate
FROM online_food_orders
GROUP BY Restaurant_Name
HAVING total_orders >= 50
ORDER BY cancellation_rate DESC
LIMIT 30;
""", engine)

cancellation_df
```

```python


plt.figure(figsize=(10,5))

plt.bar(
    cancellation_df['Restaurant_Name'],
    cancellation_df['cancellation_rate'],
    color='red'
)

plt.title("Top Restaurants by Cancellation Rate")
plt.xlabel("Restaurant Name")
plt.ylabel("Cancellation Rate (%)")
plt.xticks(rotation=45, ha='right')
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

14.Payment mode preferences

```python
payment_df = pd.read_sql("""
SELECT
    Payment_Mode,
    COUNT(Order_ID) AS total_orders,
    SUM(Final_Amount) AS total_revenue
FROM online_food_orders
GROUP BY Payment_Mode
ORDER BY total_orders DESC;
""", engine)

payment_df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))

plt.pie(
    payment_df['total_orders'],
    labels=payment_df['Payment_Mode'],
    autopct='%1.1f%%',
    startangle=140,
    colors=['skyblue', 'orange', 'green', 'purple']
)

plt.title("Payment Mode Preferences")
plt.show()
```

15.Cancellation reason analysis

```python
cancellation_reason_df = pd.read_sql("""
SELECT
    Cancellation_Reason,
    COUNT(Order_ID) AS cancelled_orders,
    (COUNT(Order_ID) / (SELECT COUNT(*) FROM online_food_orders WHERE Order_Status = 'cancelled')) * 100 AS cancellation_pct
FROM online_food_orders
WHERE Order_Status = 'cancelled'
GROUP BY Cancellation_Reason
ORDER BY cancelled_orders DESC;
""", engine)

cancellation_reason_df
```

```python


plt.figure(figsize=(8,4))

plt.bar(
    cancellation_reason_df['Cancellation_Reason'],
    cancellation_reason_df['cancelled_orders'],
    color='red'
)

plt.title("Cancellation Reasons Analysis")
plt.xlabel("Cancellation Reason")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
